## 1. Importação de Bibliotecas

Importando todas as bibliotecas necessárias para análise exploratória, visualizações e estatísticas descritivas.

```python
# Manipulação de dados
import pandas as pd
import numpy as np

# Visualização
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

# Estatística
from scipy import stats
from scipy.stats import skew, kurtosis

# Configurações
import warnings
warnings.filterwarnings('ignore')

# Configurações de visualização
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

print("✓ Bibliotecas importadas com sucesso!")
```

## 2. Carregamento dos Dados

Carregando o dataset da equipe de vendas para análise.

```python
# Carregar dados
df = pd.read_csv('equipe.csv')

# Visualizar primeiras linhas
print("="*80)
print("PRIMEIRAS LINHAS DO DATASET")
print("="*80)
display(df.head(10))

print(f"\n✓ Dataset carregado: {df.shape[0]} registros e {df.shape[1]} variáveis")
```

## 3. DICIONÁRIO DE DADOS

Documentação completa de todas as variáveis presentes no dataset, incluindo tipo, descrição e exemplos de valores únicos.

```python
# Criar dicionário de dados
dicionario_dados = pd.DataFrame({
    'Nome da Tabela': ['Equipe'] * len(df.columns),
    'Variável': df.columns,
    'Tipo': [str(df[col].dtype) for col in df.columns],
    'Tipo Estatístico': [
        'Numérica Discreta (Identificador)',
        'Qualitativa Nominal',
        'Qualitativa Nominal'
    ],
    'Descrição': [
        'Identificador único do vendedor (código numérico)',
        'Nome completo do vendedor',
        'Nome do gerente responsável pela equipe'
    ],
    'Valores Únicos': [df[col].nunique() for col in df.columns],
    'Exemplo': [df[col].iloc[0] for col in df.columns]
})

print("="*80)
print("DICIONÁRIO DE DADOS")
print("="*80)
display(dicionario_dados)

# Salvar para Excel posteriormente
print("\n✓ Dicionário de dados criado com sucesso!")
```

## 4. ANÁLISE DE QUALIDADE DOS DADOS

### Por que essa análise?
A qualidade dos dados é fundamental para garantir análises confiáveis. Identificamos valores ausentes, duplicatas e inconsistências que possam comprometer a análise.

### Análise de Valores Ausentes (Missing Values)

```python
# Análise de valores ausentes
total_registros = len(df)
missing_values = df.isnull().sum()
missing_percent = (missing_values / total_registros) * 100

qualidade_dados = pd.DataFrame({
    'Variável': df.columns,
    'Total de Registros': total_registros,
    'Valores Ausentes': missing_values,
    'Percentual Missing (%)': missing_percent.round(2)
})

print("="*80)
print("ANÁLISE DE VALORES AUSENTES")
print("="*80)
display(qualidade_dados)

print("\n📊 RESULTADO: Não há valores ausentes no dataset. Qualidade excelente!")
```

### Análise de Duplicatas

```python
# Verificar duplicatas
duplicatas_total = df.duplicated().sum()
duplicatas_id = df['id_vendedor'].duplicated().sum()

print("="*80)
print("ANÁLISE DE DUPLICATAS")
print("="*80)
print(f"Registros duplicados (todas as colunas): {duplicatas_total}")
print(f"IDs de vendedor duplicados: {duplicatas_id}")

if duplicatas_total == 0 and duplicatas_id == 0:
    print("\n✓ Não há duplicatas. Cada vendedor possui ID único.")
else:
    print("\n⚠ ATENÇÃO: Duplicatas identificadas que precisam de tratamento!")
```

### Consistência de Dados

```python
# Verificar consistência
print("="*80)
print("ANÁLISE DE CONSISTÊNCIA")
print("="*80)

# IDs únicos
ids_unicos = df['id_vendedor'].nunique()
print(f"Total de IDs únicos: {ids_unicos} de {len(df)} registros")

# Vendedores únicos
vendedores_unicos = df['nome_vendedor'].nunique()
print(f"Total de vendedores únicos: {vendedores_unicos}")

# Gerentes únicos
gerentes_unicos = df['nome_gerente'].nunique()
print(f"Total de gerentes únicos: {gerentes_unicos}")

# Verificar espaços em branco
espacos = df.applymap(lambda x: str(x).strip() != str(x) if isinstance(x, str) else False).sum()
if espacos.sum() > 0:
    print(f"\n⚠ Atenção: {espacos.sum()} campos contêm espaços extras")
else:
    print("\n✓ Não há espaços em branco desnecessários")
```

## 5. VARIÁVEIS CATEGÓRICAS - ANÁLISE DE FREQUÊNCIA

### Por que essa análise?
A análise de frequência permite entender a distribuição das categorias, identificar concentrações e possíveis desequilíbrios na base de dados.

### 5.1 Análise de Frequência: nome_vendedor

```python
# Análise de frequência - Vendedores
freq_vendedor = df['nome_vendedor'].value_counts().reset_index()
freq_vendedor.columns = ['nome_vendedor', 'Frequência Absoluta']
freq_vendedor['Frequência Absoluta Acumulada'] = freq_vendedor['Frequência Absoluta'].cumsum()
freq_vendedor['Frequência Relativa'] = (freq_vendedor['Frequência Absoluta'] / len(df) * 100).round(2)
freq_vendedor['Frequência Relativa Acumulada'] = freq_vendedor['Frequência Relativa'].cumsum().round(2)

print("="*80)
print("ANÁLISE DE FREQUÊNCIA - VENDEDORES")
print("="*80)
display(freq_vendedor)

print(f"\n📊 INTERPRETAÇÃO:")
print(f"- Total de vendedores distintos: {len(freq_vendedor)}")
print(f"- Cada vendedor aparece 1 vez (8.33% do total)")
print(f"- Não há duplicação de nomes de vendedores")
```

### 5.2 Análise de Frequência: nome_gerente

```python
# Análise de frequência - Gerentes
freq_gerente = df['nome_gerente'].value_counts().reset_index()
freq_gerente.columns = ['nome_gerente', 'Frequência Absoluta']
freq_gerente['Frequência Absoluta Acumulada'] = freq_gerente['Frequência Absoluta'].cumsum()
freq_gerente['Frequência Relativa'] = (freq_gerente['Frequência Absoluta'] / len(df) * 100).round(2)
freq_gerente['Frequência Relativa Acumulada'] = freq_gerente['Frequência Relativa'].cumsum().round(2)

print("="*80)
print("ANÁLISE DE FREQUÊNCIA - GERENTES")
print("="*80)
display(freq_gerente)

print(f"\n📊 INTERPRETAÇÃO:")
print(f"- Total de gerentes: {len(freq_gerente)}")
print(f"- Distribuição de vendedores por gerente:")
for idx, row in freq_gerente.iterrows():
    print(f"  • {row['nome_gerente']}: {row['Frequência Absoluta']} vendedores ({row['Frequência Relativa']}%)")
```

### 5.3 Visualização - Distribuição de Vendedores por Gerente

```python
# Gráfico de barras - Vendedores por Gerente
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico 1: Contagem absoluta
sns.countplot(data=df, y='nome_gerente', order=df['nome_gerente'].value_counts().index, 
              palette='viridis', ax=axes[0])
axes[0].set_title('Distribuição de Vendedores por Gerente', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Quantidade de Vendedores', fontsize=12)
axes[0].set_ylabel('Gerente', fontsize=12)

# Adicionar valores nas barras
for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%d')

# Gráfico 2: Percentual
gerente_pct = df['nome_gerente'].value_counts(normalize=True) * 100
gerente_pct.plot(kind='barh', ax=axes[1], color='skyblue')
axes[1].set_title('Percentual de Vendedores por Gerente', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Percentual (%)', fontsize=12)
axes[1].set_ylabel('Gerente', fontsize=12)

# Adicionar valores nas barras
for i, v in enumerate(gerente_pct.values):
    axes[1].text(v + 0.5, i, f'{v:.1f}%', va='center')

plt.tight_layout()
plt.savefig('distribuicao_vendedores_gerente.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Gráfico salvo como 'distribuicao_vendedores_gerente.png'")
```

### 5.4 Gráfico de Pizza - Distribuição por Gerente

```python
# Gráfico de pizza
fig, ax = plt.subplots(figsize=(10, 8))

colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']
gerente_counts = df['nome_gerente'].value_counts()

wedges, texts, autotexts = ax.pie(gerente_counts, 
                                    labels=gerente_counts.index,
                                    autopct='%1.1f%%',
                                    colors=colors,
                                    startangle=90,
                                    textprops={'fontsize': 11})

# Melhorar legibilidade
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')

ax.set_title('Distribuição Percentual de Vendedores por Gerente', 
             fontsize=14, fontweight='bold', pad=20)

# Adicionar legenda com contagens
legend_labels = [f'{name}: {count} vendedores' for name, count in gerente_counts.items()]
ax.legend(legend_labels, loc='upper left', bbox_to_anchor=(1, 1))

plt.tight_layout()
plt.savefig('distribuicao_pizza_gerente.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Gráfico salvo como 'distribuicao_pizza_gerente.png'")
```

## 6. ANÁLISE DESCRITIVA DO ID_VENDEDOR

### Por que essa análise?
Embora o id_vendedor seja um identificador, analisamos sua distribuição para verificar padrões de codificação e possíveis lacunas na numeração.

**Nota:** Não calculamos média ou outras estatísticas, pois são apenas identificadores.

```python
# Análise descritiva básica do ID
print("="*80)
print("ANÁLISE DO ID_VENDEDOR (IDENTIFICADOR)")
print("="*80)

print(f"Menor ID: {df['id_vendedor'].min()}")
print(f"Maior ID: {df['id_vendedor'].max()}")
print(f"Amplitude: {df['id_vendedor'].max() - df['id_vendedor'].min()}")
print(f"Total de IDs únicos: {df['id_vendedor'].nunique()}")

# Verificar se há sequência
ids_sorted = sorted(df['id_vendedor'].values)
print(f"\nIDs ordenados: {ids_sorted}")

# Calcular gaps
gaps = [ids_sorted[i+1] - ids_sorted[i] for i in range(len(ids_sorted)-1)]
print(f"\nMenor diferença entre IDs consecutivos: {min(gaps)}")
print(f"Maior diferença entre IDs consecutivos: {max(gaps)}")

print("\n📊 INTERPRETAÇÃO:")
print("- Os IDs não seguem uma sequência contínua (há gaps)")
print("- Isso é normal e esperado em sistemas reais onde há exclusões/inativações")
print("- Não há IDs duplicados (todos únicos)")
```

## 7. RELAÇÕES ENTRE VARIÁVEIS

### Por que essa análise?
Identificar relações entre variáveis ajuda a entender a estrutura organizacional e possíveis agrupamentos.

### 7.1 Tabela Cruzada: Gerente x Contagem de Vendedores

```python
# Tabela cruzada
tabela_cruzada = pd.crosstab(df['nome_gerente'], columns='Contagem')

print("="*80)
print("TABELA CRUZADA - GERENTE X VENDEDORES")
print("="*80)
display(tabela_cruzada)

# Estatísticas adicionais
print("\n📊 ESTATÍSTICAS POR GERENTE:")
print(f"- Média de vendedores por gerente: {tabela_cruzada['Contagem'].mean():.2f}")
print(f"- Desvio padrão: {tabela_cruzada['Contagem'].std():.2f}")
print(f"- Gerente com mais vendedores: {tabela_cruzada['Contagem'].idxmax()} ({tabela_cruzada['Contagem'].max()} vendedores)")
print(f"- Gerente com menos vendedores: {tabela_cruzada['Contagem'].idxmin()} ({tabela_cruzada['Contagem'].min()} vendedores)")
```

### 7.2 Lista Detalhada: Vendedores por Gerente

```python
# Listar vendedores por gerente
print("="*80)
print("VENDEDORES POR GERENTE (DETALHADO)")
print("="*80)

for gerente in df['nome_gerente'].unique():
    vendedores = df[df['nome_gerente'] == gerente]['nome_vendedor'].tolist()
    print(f"\n{gerente} ({len(vendedores)} vendedores):")
    for i, vendedor in enumerate(vendedores, 1):
        id_vend = df[(df['nome_gerente'] == gerente) & (df['nome_vendedor'] == vendedor)]['id_vendedor'].values[0]
        print(f"  {i}. {vendedor} (ID: {id_vend})")
```

## 8. VISUALIZAÇÕES COMPLEMENTARES

### 8.1 Heatmap de Distribuição

```python
# Criar matriz de presença (vendedor x gerente)
matriz = pd.crosstab(df['nome_vendedor'], df['nome_gerente'])

fig, ax = plt.subplots(figsize=(10, 12))
sns.heatmap(matriz, annot=True, fmt='d', cmap='YlOrRd', cbar_kws={'label': 'Presença'},
            linewidths=0.5, linecolor='gray', ax=ax)
ax.set_title('Mapa de Calor: Vendedores x Gerentes', fontsize=14, fontweight='bold', pad=20)
ax.set_xlabel('Gerente', fontsize=12)
ax.set_ylabel('Vendedor', fontsize=12)
plt.tight_layout()
plt.savefig('heatmap_vendedor_gerente.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Heatmap salvo como 'heatmap_vendedor_gerente.png'")
```

### 8.2 Treemap - Hierarquia da Equipe

## 10. EXPORTAÇÃO DOS RESULTADOS

```python
# Criar arquivo Excel com múltiplas abas
with pd.ExcelWriter('analise_exploratoria_equipe.xlsx', engine='openpyxl') as writer:
    # Aba 1: Dicionário de Dados
    dicionario_dados.to_excel(writer, sheet_name='Dicionário de Dados', index=False)
    
    # Aba 2: Dados Originais
    df.to_excel(writer, sheet_name='Dados Originais', index=False)
    
    # Aba 3: Qualidade dos Dados
    qualidade_dados.to_excel(writer, sheet_name='Qualidade dos Dados', index=False)
    
    # Aba 4: Frequência Vendedores
    freq_vendedor.to_excel(writer, sheet_name='Freq Vendedores', index=False)
    
    # Aba 5: Frequência Gerentes
    freq_gerente.to_excel(writer, sheet_name='Freq Gerentes', index=False)
    
    # Aba 6: Resumo Executivo
    resumo = pd.DataFrame({
        'Métrica': ['Total de Vendedores', 'Total de Gerentes', 'Média Vendedores/Gerente',
                   'Valores Ausentes', 'Duplicatas', 'Qualidade Geral'],
        'Valor': [len(df), df['nome_gerente'].nunique(), 
                 f"{len(df) / df['nome_gerente'].nunique():.2f}",
                 '0 (0%)', '0', 'Excelente']
    })
    resumo.to_excel(writer, sheet_name='Resumo Executivo', index=False)

print("\n" + "="*80)
print("✓ Análise concluída com sucesso!")
print("✓ Arquivo Excel gerado: analise_exploratoria_equipe.xlsx")
print("✓ Gráficos salvos em formato PNG e HTML")
print("="*80)
```
